# Arbitrary-precision evaluation of the CDF $F(\tau, \mathrm{C}, \mathrm{Q})$

Example code which shows how to compute the CDF $F(\tau, \mathrm{C}, \mathrm{Q})$ for several different filters $\mathrm{Q}$ and sets of pulsars, as described in the subsection *Numerical evaluation of the CDF* of the paper.

All calculations are done by `compute_cdf.py`; this notebook only calls it.

Requires `numpy`, `matplotlib` and `mpmath`. `mpmath` is much faster with the optional `gmpy2` backend; without it, the plotting cells below take tens of minutes.

```python
import compute_cdf as cc

# check that we are using an efficient arbitrary-precision math library
cc.check_mpmath_backend()
```

```python
# make dictionary of eigenvalues of E for the NP, NPMV and DF filters, for different sets of pulsars
Evals = {key: cc.make_eigenvalue_dict(filename) for key, filename in cc.PULSAR_FILES.items()}

# print number of NPMV eigenvalues of each sign
for key,evals in Evals.items():
    print(f' ========= {key} pulsars ============')
    cc.print_eigenvalue_signs(evals['EVNnpmv'])
```

```python
# Illustrate precision. Change values of tau as desired
tau = -10.0
for key,evals in Evals.items():
    print(f'\n{key} PULSAR CASE')
    cc.print_precision_test(tau, evals['EVNnpmv'])
```

```python
# Plot CDF for the different sets of eigenvalues, showing mantissa bits needed
for key,evals in Evals.items():
    for key2, ev in evals.items():
        cc.plot_cdf(ev,f'{key2} with {key}')
```

```python
# plot ROC for the different PTAs
for case,evals in Evals.items():
    print(f'Doing ROC curves for {case} pulsars')
    cc.plot_all_ROC(evals, case)
```

```python
# print detection probabilities (DP) at the 5-sigma false-alarm probability (FAP)
for case,evals in Evals.items():
    print("==================================")
    cc.find_tau_for_fap(evals, case, reltol=1.e-5)
```
